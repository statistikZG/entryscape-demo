```python
from rdflib import Graph
```

```python
catalog_url = "https://zg-demo.entryscape.net/store/1"
resource_id = "7"
```

```python

g = Graph()
# Parse dataset RDF
dataset_url = f"{catalog_url}/resource/{resource_id}"
g.parse(dataset_url)
g.print(format="turtle")
```
